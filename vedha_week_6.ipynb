```python
df=df.drop_duplicates()
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
#Gender distribution


sns.countplot(data=df, x="Gender")
plt.title("Patient Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of patients")
plt.show()
```

```python
#Age group distribution

age_bins = [-1, 17, 39, 59, float("inf")]

age_labels = [
    "Child",
    "Young Adults",
    "Middle Adults",
    "Senior "
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins = age_bins,
    labels = age_labels
)
```

```python
df["Age Group"].value_counts()
```

```python
sns.countplot(data=df, x="Age Group")
plt.title("Patients Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of patients")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Billing Amount"

    )

plt.title("Age vs Billing Amount")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.show()
```

```python
#understanding admission data

df["Date of Admission"].head()
```

```python
df["Date of Admission"].dtype
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df[
    [
        "Date of Admission",
        "Admission Year",
        "Admission Month"
    ]
].head()
```

```python
#Count Monthly Admissions

monthly_admissions = (
    df.groupby("Admission Month")
      .size()
      .reset_index(name="Admissions")
)
```

```python
monthly_admissions
```

```python
#Admission trend

sns.lineplot(data=monthly_admissions, x="Admission Month", y="Admissions",markers = "o")

plt.title("Monthly Admissions Trend")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation = 45)

plt.show()
```

```python
monthly_admissions_type = df.groupby(["Admission Year", "Admission Month"])["Admission Type"].value_counts().reset_index(name="Admissions")
```

```python
monthly_admissions_type.head()
```

```python
sns.lineplot(data=monthly_admissions_type, x="Admission Month", y="Admissions", hue="Admission Type", markers = "o")

plt.title("Monthly Admissions by Admission Type")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation = 45)

plt.show()
```

```python
df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python
sns.scatterplot(data=df, x="Length of Stay", y="Billing Amount")
plt.title("Length of Stay vs Billing Amount")
plt.xlabel("Length of Stay")
plt.ylabel("Billing Amount")
plt.show()
```

```python
numeric_cols = [
    "Age",
    "Length of Stay",
    "Billing Amount"
]

corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1)

plt.title("Healthcare variables correlation matrix")
plt.show()
```
